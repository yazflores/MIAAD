```python
"""
Complete únicamente las cinco expresiones marcadas con TODO. No cambie la
firma de la función, las claves del diccionario ni los textos de salida.
"""


def evaluar_decision(
    produccion,
    consumo_observado,
    capacidad_contratada,
    intercepto=74.8,
    pendiente=0.153,
    produccion_min=420.0,
    produccion_max=1180.0,
    costo_exceso=45.0,
    costo_deficit=220.0,
):
    """Evalúa una predicción y una decisión de contratación ya tomada."""

    # TODO 1: calcule la predicción de la línea base.
    prediccion = intercepto + pendiente * produccion

    # TODO 2: calcule el residuo como observado menos predicho.
    residuo = consumo_observado - prediccion

    # TODO 3: sustituya False por la comparación que incluya ambos límites.
    if produccion_min <= produccion <= produccion_max:
        alcance = "interpolacion"
    else:
        alcance = "extrapolacion"

    diferencia_capacidad = capacidad_contratada - consumo_observado

    if diferencia_capacidad > 0:
        situacion = "exceso"
        magnitud = diferencia_capacidad
        # TODO 4: calcule el costo del exceso.
        costo = magnitud * costo_exceso
    elif diferencia_capacidad < 0:
        situacion = "deficit"
        magnitud = -diferencia_capacidad
        # TODO 5: calcule el costo del déficit.
        costo = magnitud * costo_deficit
    else:
        situacion = "exacta"
        magnitud = 0.0
        costo = 0.0

    return {
        "prediccion_mwh": round(prediccion, 2),
        "residuo_mwh": round(residuo, 2),
        "alcance": alcance,
        "situacion": situacion,
        "magnitud_mwh": round(magnitud, 2),
        "costo_mxn": round(costo, 2),
    }
```
